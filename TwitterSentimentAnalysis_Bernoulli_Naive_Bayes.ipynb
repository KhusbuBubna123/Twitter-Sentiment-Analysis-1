#Text Preprocessing.

#Removing punctuation and converting to lower case.

```python
def remove_punct(text):
    return ''.join(ch for ch in text if ch not in string.punctuation)

def convert_lower(text):
    return text.lower()

df['Tweet_clean'] = df['Tweet'].apply(remove_punct)
df['Tweet_clean'] = df['Tweet_clean'].apply(convert_lower)
df.head(8)
```

#Splitting into Train and Test Data

```python
X = df['Tweet_clean'].values
Y = df['Sentiment'].values.astype(int)

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.20, random_state=10)

print('Train size:', len(X_train))
print('Test  size:', len(X_test))
```

#Vectorisation — Bag of Words (CountVectorizer)

```python
#Count Vectorize and remove stopwords.
CV = CountVectorizer(stop_words='english')
X_train_cv = CV.fit_transform(X_train)
X_test_cv  = CV.transform(X_test)

print('Vocabulary size:', len(CV.vocabulary_))
print('Train shape    :', X_train_cv.shape)
print('Test  shape    :', X_test_cv.shape)
```

```python
feature_names = CV.get_feature_names_out()
print('Total features:', len(feature_names))
print('Sample feature tokens  :', feature_names[:100]) #Printing First 100 tokens from the Vocaborary.
```

#Training the model — Bernoulli Naive Bayes

```python
NB = BernoulliNB(alpha=0.01) #alpha is used for Laplace smooothing.
NB.fit(X_train_cv, Y_train)
Y_Predict = NB.predict(X_test_cv)
Y_Predict
```

```python
#y_predict=y_predict.replace({0:"beningn",1:"Malignant"})
y_cat_variable = pd.Categorical(Y_Predict).rename_categories({0:"Negative Sentiment", 4:"Positive Sentiment"})
y_cat_variable

#(0 = negative, 4 = positive)
```

#Accuracy

```python
print(f'Bernoulli Naive Bayes Accuracy using Count Vectorizer: {accuracy_score(Y_test, Y_Predict)*100:.2f}%')
```

#F1 score

```python
f1_score=f1_score(Y_Predict,Y_test,pos_label=4)
f1_score
```

```python
recall=recall_score(Y_Predict,Y_test,pos_label=4)
recall
```

```python
print(classification_report(Y_test, Y_Predict, target_names=['Positive Sentiment', 'Negative Sentiment']))
```

#Printing Confusion Matrix.

```python
cm = pd.DataFrame(
    confusion_matrix(Y_test, Y_Predict),
    index=['Actual Positive', 'Actual Negative'],
    columns=['Predicted Positive', 'Predicted Negative'],
)
cm
```

#Testing on a New Tweet.

```python
def classify_tweet(tweet):
    cleaned  = remove_punct(tweet)
    cleaned = convert_lower(cleaned)
    features = CV.transform([cleaned])
    Predict = NB.predict(features)[0]
    if(Predict == 0):
      Sentiment='Negative Sentiment'
    elif(Predict== 4):
      Sentiment='Positive Sentiment'
    return Sentiment
#(0 = negative, 4 = positive)
```
